```python
# Find-S Algorithm in Google Colab

import pandas as pd

# Create dataset
data = pd.DataFrame([
    ['Good','Good','Yes','Good','Yes'],
    ['Good','Average','Yes','Good','Yes'],
    ['Average','Good','Yes','Average','Yes'],
    ['Poor','Good','No','Good','No'],
    ['Good','Poor','No','Average','No']
], columns=['Programming','Academics','Project','Communication','Eligible'])

print("Dataset:")
print(data)

# Select only positive examples
positive_examples = data[data['Eligible'] == 'Yes'].iloc[:, :-1].values

# Initialize hypothesis with first positive example
hypothesis = positive_examples[0].copy()

# Apply Find-S algorithm
for example in positive_examples:
    for i in range(len(hypothesis)):
        if hypothesis[i] != example[i]:
            hypothesis[i] = '?'

print("\nFinal Hypothesis using Find-S Algorithm:")
print(hypothesis)
```

```python

# Improved Candidate Elimination Algorithm (Pruned G)

def candidate_elimination(examples, labels):
    S = examples[0].copy()
    G = [['?' for _ in range(len(S))]]

    for i, example in enumerate(examples):

        if labels[i] == 'Yes':  # Positive example
            for j in range(len(S)):
                if S[j] != example[j]:
                    S[j] = '?'

            G = [g for g in G
                 if all(g[j] == '?' or g[j] == example[j]
                        for j in range(len(S)))]

        else:  # Negative example
            new_G = []
            for g in G:
                for j in range(len(S)):
                    if g[j] == '?' and S[j] != example[j]:
                        h = g.copy()
                        h[j] = S[j]
                        if h not in new_G:
                            new_G.append(h)
            G = new_G

    # Remove overly general hypotheses
    G = [g for g in G if g != ['?' for _ in range(len(S))]]

    return S, G


X = data.iloc[:, :-1].values
y = data['Eligible'].values

S, G = candidate_elimination(X, y)

print("Specific Boundary (S):", S)
print("General Boundary (G):", G)

```
